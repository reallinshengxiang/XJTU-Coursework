```python
import re
import matplotlib.pyplot as plt
import os
import matplotlib.font_manager as fm
import numpy as np

def parse_result_file(filename):
    """解析Cache模拟器结果文件，提取参数和性能指标"""
    print(f"\n正在解析文件: {filename}")
    try:
        with open(filename, 'r', encoding='GB2312') as f:
            text = f.read()
    except Exception as e:
        print(f"读取文件失败: {e}")
        return []
    section_pattern = re.compile(r'-{5,}\s*([^\n]+)\s*-{5,}([\s\S]*?)(?=-{5,}|$)')
    sections = section_pattern.findall(text)
    print(f"找到 {len(sections)} 个带标题的部分")
    param_sections = []
    result_sections = []
    
    for title, content in sections:
        if "Cache Simulator输入参数" in title:
            param_sections.append(content)
        elif "Cache 基本性能分析结果" in title:
            result_sections.append(content)
    
    print(f"找到 {len(param_sections)} 个参数部分 和 {len(result_sections)} 个结果部分")
    
    if len(param_sections) != len(result_sections):
        print(f"警告: 参数部分和结果部分数量不匹配: {len(param_sections)} vs {len(result_sections)}")
        min_len = min(len(param_sections), len(result_sections))
        param_sections = param_sections[:min_len]
        result_sections = result_sections[:min_len]  
    data_list = []
    for i, (param_text, result_text) in enumerate(zip(param_sections, result_sections)):
        param_lines = [line.strip() for line in param_text.split('\n') if line.strip()]
        input_file = None
        for line in param_lines:
            if line.startswith("输入文件："):
                input_file = line.split("：", 1)[1].strip()
                input_file = os.path.basename(input_file)
                break
        if not input_file:
            print(f"警告: 第 {i+1} 部分未找到输入文件名")
            continue
        cache_size = None
        block_size = None
        associativity = None
        for line in param_lines:
            if "Cache大小" in line and ("块大小" in line or "行大小" in line) and ("相联度" in line or "路" in line):
                # 尝试多种匹配模式
                patterns = [
                    r'同一的Cache大小\s*[:：]\s*(\d+)\s*KB\s*块大小\s*[:：]\s*(\d+)\s*B\s*相联度\s*[:：]\s*(\d+)\s*路',
                    r'Cache大小\s*[:：]\s*(\d+)\s*KB\s*块大小\s*[:：]\s*(\d+)\s*B\s*相联度\s*[:：]\s*(\d+)\s*路',
                    r'Cache大小\s*(\d+)\s*KB\s*块大小\s*(\d+)\s*B\s*相联度\s*(\d+)\s*路',
                ]
                
                for pattern in patterns:
                    match = re.search(pattern, line)
                    if match:
                        cache_size = int(match.group(1))
                        block_size = int(match.group(2))
                        associativity = int(match.group(3))
                        break
                
                if cache_size is not None:
                    break
        
        if cache_size is None or block_size is None or associativity is None:
            print(f"警告: 第 {i+1} 部分({input_file}) 未找到完整的Cache参数")
            continue
        result_lines = [line.strip() for line in result_text.split('\n') if line.strip()]
        metrics = {
            'total': None,
            'instruction': None,
            'data_read': None,
            'data_write': None
        }
        for line in result_lines:
            line = line.strip()
            if line.startswith("访问总次数") and "不命中率" in line:
                match = re.search(r'不命中率\s*[:：]\s*([\d.]+)\s*%', line)
                if match:
                    metrics['total'] = float(match.group(1))
            elif line.startswith("读指令次数") and "不命中率" in line:
                match = re.search(r'不命中率\s*[:：]\s*([\d.]+)\s*%', line)
                if match:
                    metrics['instruction'] = float(match.group(1))
            elif line.startswith("读数据次数") and "不命中率" in line:
                match = re.search(r'不命中率\s*[:：]\s*([\d.]+)\s*%', line)
                if match:
                    metrics['data_read'] = float(match.group(1))
            elif line.startswith("写数据次数") and "不命中率" in line:
                match = re.search(r'不命中率\s*[:：]\s*([\d.]+)\s*%', line)
                if match:
                    metrics['data_write'] = float(match.group(1))
        if all(value is None for value in metrics.values()):
            print(f"警告: 第 {i+1} 部分({input_file}) 未找到任何性能指标")
            continue
        data_list.append({
            'input_file': input_file,
            'cache_size': cache_size,
            'block_size': block_size,
            'associativity': associativity,
            **metrics
        })
    print(f"\n成功解析 {len(data_list)} 条数据")
    return data_list

def plot_analysis(data):
    """分析并绘制Cache性能数据"""
    if not data:
        print("Error: No data available for plotting")
        return
    files = sorted({d['input_file'] for d in data})
    if not files:
        print("Error: No input files identified")
        return
    print(f"\n开始分析 {len(files)} 个输入文件的数据")
    
    for file in files:
        file_data = [d for d in data if d['input_file'] == file]
        print(f"\n分析文件: {file} ({len(file_data)} 条记录)")
        cache_sizes = sorted({d['cache_size'] for d in file_data})
        block_sizes = sorted({d['block_size'] for d in file_data})
        associativities = sorted({d['associativity'] for d in file_data})
        print(f"  Cache大小: {cache_sizes} KB")
        print(f"  块大小: {block_sizes} B")
        print(f"  相联度: {associativities} 路")
        plt.figure(figsize=(15, 10))
        plt.suptitle(f"File Analysis: {file}\nEffect of Block Size on Miss Rate for Fixed Cache Sizes", y=1.02)
        display_cache_sizes = min(4, len(cache_sizes))
        if display_cache_sizes == 0:
            print(f"Warning: Insufficient cache size data for {file}")
            continue
        for idx, cache_size in enumerate(cache_sizes[:display_cache_sizes]):
            ax = plt.subplot(2, 2, idx + 1)
            subset = [d for d in file_data 
                     if d['cache_size'] == cache_size 
                     and d['associativity'] in associativities]
            if not subset:
                print(f"Warning: No valid data for {cache_size}KB cache")
                ax.text(0.5, 0.5, f"No Data: {cache_size}KB", ha='center', va='center')
                continue
            print(f"  绘制Cache大小 {cache_size}KB: {len(subset)} 条数据")
            for assoc in associativities:
                assoc_data = [d for d in subset if d['associativity'] == assoc]
                if not assoc_data:
                    continue
                assoc_data = sorted(assoc_data, key=lambda x: x['block_size'])
                x = [d['block_size'] for d in assoc_data]
                y = [d['total'] for d in assoc_data]
                print(f"    {assoc}路: 块大小={x}, 不命中率={y}")
                ax.plot(x, y, marker='o', label=f'{assoc}-Way')
            ax.set_title(f'Cache Size: {cache_size}KB')
            ax.set_xlabel('Block Size (Bytes)')
            ax.set_ylabel('Total Miss Rate (%)')
            ax.set_xscale('log', base=2)
            ax.set_xticks(x)
            ax.set_xticklabels([str(bs) for bs in x])
            ax.grid(True)
            ax.legend()
        plt.tight_layout()
        plt.show()
        plt.figure(figsize=(15, 10))
        plt.suptitle(f"File Analysis: {file}\nEffect of Associativity on Miss Rate for Fixed Block Sizes", y=1.02)
        display_block_sizes = min(4, len(block_sizes))
        if display_block_sizes == 0:
            print(f"Warning: Insufficient block size data for {file}")
            continue
        for idx, block_size in enumerate(block_sizes[:display_block_sizes]):
            ax = plt.subplot(2, 2, idx + 1)
            subset = [d for d in file_data 
                     if d['block_size'] == block_size 
                     and d['cache_size'] in cache_sizes]
            if not subset:
                print(f"Warning: No valid data for {block_size}B blocks")
                ax.text(0.5, 0.5, f"No Data: {block_size}B", ha='center', va='center')
                continue
            print(f"  绘制块大小 {block_size}B: {len(subset)} 条数据")
            for cache_size in cache_sizes:
                cache_data = [d for d in subset if d['cache_size'] == cache_size]
                if not cache_data:
                    continue
                cache_data = sorted(cache_data, key=lambda x: x['associativity'])
                x = [d['associativity'] for d in cache_data]
                y = [d['total'] for d in cache_data]
                print(f"    {cache_size}KB: 相连度={x}, 不命中率={y}")
                ax.plot(x, y, marker='s', label=f'{cache_size}KB')
            ax.set_title(f'Block Size: {block_size}B')
            ax.set_xlabel('Associativity (Ways)')
            ax.set_ylabel('Total Miss Rate (%)')
            ax.set_xscale('log', base=2)
            ax.set_xticks(x)
            ax.set_xticklabels([str(a) for a in x])
            ax.grid(True)
            ax.legend()
        plt.tight_layout()
        plt.show()
        plt.figure(figsize=(15, 10))
        plt.suptitle(f"File Analysis: {file}\nMiss Rate Comparison by Access Type for Varying Cache Sizes", y=1.02)
        common_block_size = max(set(d['block_size'] for d in file_data), 
                              key=lambda x: sum(1 for d in file_data if d['block_size'] == x))
        common_associativity = max(set(d['associativity'] for d in file_data), 
                                 key=lambda x: sum(1 for d in file_data if d['associativity'] == x))
        print(f"  使用常见块大小: {common_block_size}B, 相连度: {common_associativity}路")
        subset = [d for d in file_data 
                 if d['block_size'] == common_block_size 
                 and d['associativity'] == common_associativity]
        if not subset:
            print(f"Warning: No valid data for block size={common_block_size}B, associativity={common_associativity} ways")
        else:
            subset = sorted(subset, key=lambda x: x['cache_size'])
            x = [d['cache_size'] for d in subset]
            instruction_miss = [d['instruction'] for d in subset]
            data_read_miss = [d['data_read'] for d in subset]
            data_write_miss = [d['data_write'] for d in subset]
            total_miss = [d['total'] for d in subset]
            plt.plot(x, instruction_miss, marker='o', label='Instruction Miss Rate')
            plt.plot(x, data_read_miss, marker='s', label='Data Read Miss Rate')
            plt.plot(x, data_write_miss, marker='^', label='Data Write Miss Rate')
            plt.plot(x, total_miss, marker='D', label='Total Miss Rate')
            plt.title(f'Block Size: {common_block_size}B, Associativity: {common_associativity} Ways')
            plt.xlabel('Cache Size (KB)')
            plt.ylabel('Miss Rate (%)')
            plt.xscale('log', base=2)
            plt.xticks(x, [str(s) for s in x])
            plt.grid(True)
            plt.legend()
            plt.tight_layout()
            plt.show()
        plt.figure(figsize=(15, 10))
        plt.suptitle(f"File Analysis: {file}\nMiss Rate Comparison by Access Type for Varying Associativity", y=1.02)
        common_cache_size = max(set(d['cache_size'] for d in file_data), 
                              key=lambda x: sum(1 for d in file_data if d['cache_size'] == x))
        print(f"  使用常见Cache大小: {common_cache_size}KB, 块大小: {common_block_size}B")
        subset = [d for d in file_data 
                 if d['cache_size'] == common_cache_size 
                 and d['block_size'] == common_block_size]
        if not subset:
            print(f"Warning: No valid data for cache size={common_cache_size}KB, block size={common_block_size}B")
        else:
            subset = sorted(subset, key=lambda x: x['associativity'])
            x = [d['associativity'] for d in subset]
            instruction_miss = [d['instruction'] for d in subset]
            data_read_miss = [d['data_read'] for d in subset]
            data_write_miss = [d['data_write'] for d in subset]
            total_miss = [d['total'] for d in subset]
            plt.plot(x, instruction_miss, marker='o', label='Instruction Miss Rate')
            plt.plot(x, data_read_miss, marker='s', label='Data Read Miss Rate')
            plt.plot(x, data_write_miss, marker='^', label='Data Write Miss Rate')
            plt.plot(x, total_miss, marker='D', label='Total Miss Rate')
            plt.title(f'Cache Size: {common_cache_size}KB, Block Size: {common_block_size}B')
            plt.xlabel('Associativity (Ways)')
            plt.ylabel('Miss Rate (%)')
            plt.xscale('log', base=2)
            plt.xticks(x, [str(a) for a in x])
            plt.grid(True)
            plt.legend()
            plt.tight_layout()
            plt.show()
        plt.figure(figsize=(15, 10))
        plt.suptitle(f"File Analysis: {file}\nMiss Rate Comparison by Access Type for Varying Block Sizes", y=1.02)
        subset = [d for d in file_data 
                 if d['cache_size'] == common_cache_size 
                 and d['associativity'] == common_associativity]
        if not subset:
            print(f"Warning: No valid data for cache size={common_cache_size}KB, associativity={common_associativity} ways")
        else:
            subset = sorted(subset, key=lambda x: x['block_size'])
            x = [d['block_size'] for d in subset]
            instruction_miss = [d['instruction'] for d in subset]
            data_read_miss = [d['data_read'] for d in subset]
            data_write_miss = [d['data_write'] for d in subset]
            total_miss = [d['total'] for d in subset]
            plt.plot(x, instruction_miss, marker='o', label='Instruction Miss Rate')
            plt.plot(x, data_read_miss, marker='s', label='Data Read Miss Rate')
            plt.plot(x, data_write_miss, marker='^', label='Data Write Miss Rate')
            plt.plot(x, total_miss, marker='D', label='Total Miss Rate')
            plt.title(f'Cache Size: {common_cache_size}KB, Associativity: {common_associativity} Ways')
            plt.xlabel('Block Size (Bytes)')
            plt.ylabel('Miss Rate (%)')
            plt.xscale('log', base=2)
            plt.xticks(x, [str(bs) for bs in x])
            plt.grid(True)
            plt.legend()
            plt.tight_layout()
            plt.show()

def print_file_preview(filename, lines=20):
    """打印文件前几行用于调试"""
    try:
        with open(filename, 'r', encoding='GB2312') as f:
            print(f"\n文件 {filename} 的前 {lines} 行预览:")
            for i, line in enumerate(f):
                if i >= lines:
                    break
                print(f"{i+1}: {line.strip()}")
    except Exception as e:
        print(f"预览文件失败: {e}")

if __name__ == "__main__":
    print_file_preview('/kaggle/input/result/result.txt')
    data = parse_result_file('/kaggle/input/result/result.txt')
    plot_analysis(data)
```
